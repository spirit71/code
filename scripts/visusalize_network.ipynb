```python
import network
import torch
from config_parser import parse_arguments
import random
import numpy as np
# --- 0. 设置确定性 ---
SEED = 42
torch.manual_seed(SEED)
torch.cuda.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
np.random.seed(SEED)
random.seed(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False 
print(f"设定随机种子为 {SEED}，确保追踪过程的确定性。")
# --- 检查 GPU 可用性 ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if device.type == 'cpu':
    print("⚠️ 警告：检测到 CPU。xFormers 算子无法在 CPU 上导出，请在 GPU 环境下运行。")
    # 如果没有 GPU，您必须修改 network.py 禁用 xFormers。
# --- 1. 参数解析与路径修正 ---
args = parse_arguments()

# ⚠️ 基础模型路径修正 (假设您已在config_parser中修正或确保路径正确，这里仅为示例)
# 确保 args.foundation_model_path 指向正确的 .pth 文件
args.foundation_model_path = "../weights/dinov2_vitb14_pretrain.pth" 
print(f"基础模型路径: {args.foundation_model_path}")

#### Initialize model (加载基础模型权重，这会调用 network.py 中的 torch.load)
model = network.VPRNet(pretrained_foundation=True, foundation_model_path=args.foundation_model_path)
# --- 2. 关键：将模型移动到 GPU ---
model.to(device)
model.half()
# --- 3. 加载主模型权重并修正 DataParallel (DP) ---
state_dict = torch.load('../weights/best_model.pth', weights_only=False, map_location=device) 
new_state_dict = {k.replace('module.', ''): v for k, v in state_dict.items()}

# # 假设 best_model.pth 是在 DataParallel 下保存的，需要移除 'module.' 前缀
print("正在移除 DataParallel 前缀...")
new_state_dict = {k.replace('module.', ''): v for k, v in state_dict.items()}

# 加载到原始模型
model.load_state_dict(new_state_dict,strict=False) 
 
model.eval() # 必须设置为评估模式

# ⚠️ 关键修正：删除或注释掉 model = torch.nn.DataParallel(model) 这一行!
# ONNX 导出必须使用原始的 model 实例！
# model = torch.nn.DataParallel(model) 
# ⚡ 新增：临时禁用模型中的所有 Dropout 层
# 确保 tracing sanity checks 通过
for module in model.modules():
    if isinstance(module, torch.nn.Dropout):
        # 记录原始概率，以便后续恢复 (尽管这里不需要恢复，但这是好习惯)
        module._original_p = module.p
        module.p = 0.0
        print(f"临时禁用 Dropout 层: 设置 p={module._original_p} 为 p=0.0")
dummy_input = torch.randn(1, 3, 224, 224, dtype=torch.float16).to(device)

if device.type == 'cuda':
     torch.cuda.empty_cache()

output_path = "jit_exported_model.pt"

print(f"\n--- 尝试 Scripting (推荐用于复杂的Transformer) ---")
try:
    scripted_model = torch.jit.script(model)
    scripted_model.save(output_path)
    print(f"✅ 模型脚本化和保存成功！文件名为：{output_path}")
    print("请用 Netron (https://netron.app/) 打开该文件。")

except Exception as script_e:
    print(f"❌ Scripting 失败！错误信息：{script_e}")
    print(f"--- 尝试回退到 Tracing (使用严格模式) ---")

    try:
        # 尝试使用 torch.jit.trace 追踪模型 (保持默认 strict=True)
        traced_script_module = torch.jit.trace(model, dummy_input,strict=False)
        traced_script_module.save(output_path)
        print(f"✅ 模型追踪和保存成功！文件名为：{output_path}")
        print("请用 Netron (https://netron.app/) 打开该文件。")
        
    except Exception as trace_e:
        # 如果 Tracing 也失败，则打印详细信息
        print(f"\n❌ 模型导出最终失败！")
        print(f"Scripting 失败原因：{script_e}")
        print(f"Tracing 失败原因：{trace_e}")
        print("\n**故障排除提示：**")
        print("* 1. 如果 Tracing 依然失败，请尝试在 `torch.jit.trace` 中添加 `strict=False`，但请注意模型可能不正确。")
        print("* 2. 考虑彻底禁用 `xFormers`，并在模型中注释掉任何可能导致动态行为的自定义逻辑。")

# # --- 3. ONNX 导出 ---
# dummy_input = torch.randn(1, 3, 224, 224, dtype=torch.float16).to(device)
# if device.type == 'cuda':
#     torch.cuda.empty_cache() # 释放缓存
# print("开始导出 ONNX...")
# torch.onnx.export(
#     model, 
#     dummy_input, 
#     "model.onnx", 
#     export_params=True, 
#     opset_version=17, 
#     do_constant_folding=True, 
#     input_names=['input'], 
#     output_names=['output'], 
# )
# print("✅ ONNX 导出成功！请用 Netron 打开 model.onnx。")
```
